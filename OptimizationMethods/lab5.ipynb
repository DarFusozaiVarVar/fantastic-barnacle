```python
import numpy as np
import sympy as sp
import pandas as pd
import plotly.graph_objects as go
import random 

def rand_crd_s_stepb(f, x0, lamb = 0.5, alpha = 2, eps = 0.1, maxiter = 1000, n = 2):
    
    x = x0
    fun = sp.lambdify([x1, x2], f, "math")
    m = 1
    e_list = [np.array([1, 0]),np.array([0, 1])]
    sk = 1
    x_path = [x]
    f_path = [fun(x[0], x[1])]
    

    for k in range(0, maxiter):
        i = k % n
        p = e_list[i]
        x_new = x + alpha*p
        fx = fun(x[0], x[1])
        fx_new = fun(x_new[0], x_new[1])

        if fx > fx_new:
            x_path.append(x_new)
            f_path.append(fx_new)
            continue

        x_new = x - alpha*p
        fx_new = fun(x_new[0], x_new[1])
        
        if fx > fx_new:
            x_path.append(x_new)
            f_path.append(fx_new)
            continue

        if i == n:
            alpha *= lamb

        if i == n and k >= n:
            if np.norm(x-x_new) < eps:
                x_path.append(x_new)
                f_path.append(fx_new)


    return np.array(x_path), np.array(f_path), k+1


x1, x2 = sp.symbols('x1, x2')
f = 100*(1 - x1**2 - x2**2)**2 + (x1-1)**2 + x2**2
np.random.seed(42)
p_list = (np.random.uniform(-3, 3, size=2), np.random.uniform(-3, 3, size=2), np.random.uniform(-3, 3, size=2))

x1_range = np.linspace(-10, 10, 100)
x2_range = np.linspace(-10, 10, 100)
x, y = np.meshgrid(x1_range, x2_range)
fu = 100*(1 - x**2 - y**2)**2 + (x-1)**2 + y**2
fig = go.Figure(data=[go.Contour(
    zmax = 5000, zmin = 0,
    z=fu, x=x1_range, y=x2_range, colorscale='Jet',
    contours=dict(
         coloring = 'heatmap',
        showlabels = True,
        labelfont = dict(
            size = 12,
            color = 'white',
        )
    ) 
)])
fig.update_layout(title=f'100(1-x1^2-x2^2)^2+(x1-1)^2+x2^2') 

# method visualize

for point in p_list:
    print(f'point = {point}')
    ans = rand_crd_s_stepb(f, point)
    
    print(ans[0][-1], ans[1][-1], ans[2])
    
    fig.add_trace(go.Scatter(x=[i[0] for i in ans[0]], y=[i[1] for i in ans[0]], name=f'точка {point.tolist()}'))
    fig.update_layout(
    legend=dict(
        orientation="h",   
        yanchor="bottom",
        y=1.02,            
        xanchor="right",
        x=1               
    )
)
fig.show()
```
