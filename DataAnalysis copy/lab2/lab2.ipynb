## Анализ данных карьерного самосвала
- Для выделения ям будем использовать преобразование Фурье, как алгоритм фильтрации данных.

```python
import math 
import pandas as pd
import numpy as np
from scipy.fft import rfft, rfftfreq, irfft
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Тут мы посмотрим как делается преобразование фурье на простом примере с двумя синусоидами.

freq = 1
interval = 5
n_points = 200
x = [t/n_points for t in range(int(n_points*interval))]
y = [2*np.sin(2*np.pi*freq*t) + 0.3*np.sin(2*np.pi*10*freq*t) for t in x]

fig = make_subplots(specs=[[{"secondary_y": True}]])
fig.add_trace(go.Scatter(x=x, y=y, mode='lines', name='Гармоническая функция'), secondary_y=True)
fig.update_layout(title='Функция', width=1100, height=500)
fig.show()

yf = rfft(y)
abs_yf = [np.abs(i)/len(yf) for i in yf]
xf = rfftfreq(len(y), 1/n_points)

fig = make_subplots(specs=[[{"secondary_y": True}]])
fig.add_trace(go.Bar(x=xf[:100], y=abs_yf[:100], width=.1, name='Spectr'), secondary_y=False)
fig.update_layout(title='Спектр', width=1200, height=500)
fig.show()

y = irfft(yf)

fig = make_subplots(specs=[[{"secondary_y": True}]])
fig.add_trace(go.Scatter(x=x, y=y, mode='lines', name='Гармоническая функция'), secondary_y=True)
fig.update_layout(title='Функция после обратного преобразования', width=1100, height=500)
fig.show()
```
